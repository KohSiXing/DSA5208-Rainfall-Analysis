```python
# Cell 1: am I on the cluster?
import socket
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
print(socket.gethostname(), spark.sparkContext.master, spark.version)
```

```python
# Cell 2: read the rainfall data
from pyspark.sql import functions as F

DATA = "gs://dsa5208-rainfall-ashley-data/raw/rainfall_*.csv"
rain = spark.read.option("header", True).csv(DATA)
rain.printSchema()
rain.show(5, truncate=False)
```

```python
# Cell 3: how many rows, and per file?
print("Total rows:", rain.count())
(rain.withColumn("file", F.input_file_name())
     .groupBy("file").count()
     .orderBy("file")
     .show(truncate=False))
```
